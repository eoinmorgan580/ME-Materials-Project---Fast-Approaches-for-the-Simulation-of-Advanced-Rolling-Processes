```python
import pyroll.basic as pr
```

```python
in_profile = pr.Profile.round(
    diameter = 20e-3,
    temperature = 298.15,
    density = 7850,
    specific_heat_capacity = 500,
    freiberg_flow_stress_coefficients = pr.FreibergFlowStressCoefficients(
        a = 900e6,
        m1 = 0,
        m2 = 0.25,
        m3 = 0,
        m4 = 0,
        m5 = 0,
        m6 = 0,
        m7 = 0,
        m8 = 0,
        m9 = 0,
        baseStrain = 1,
        baseStrainRate = 0
    )
)

in_profile
```

```python
groove1 = pr.BoxGroove(
    r1 = 1e-3,
    r2 = 1e-3,
    depth = 1e-3,
    ground_width = 30e-3,
    usable_width = 32e-3
)

groove1
```

```python
roll1 = pr.Roll(
    groove = groove1,
    nominal_radius = 200e-3
)

roll1
```

```python
pass1 = pr.RollPass(
    label = 'Flat Pass 1',
    roll = roll1,
    gap = 14e-3,
    velocity = 1.586
)

pass1
```

```python
pass2 = pr.RollPass(
    label = 'Flat Pass 2',
    roll = roll1,
    gap = 13e-3,
    velocity = 1.622,
    rotation = 0
)

pass2
```

```python
pass3 = pr.RollPass(
    label = 'Flat Pass 3',
    roll = roll1,
    gap = 12e-3,
    velocity = 1.667,
    rotation = 0
)

pass3
```

```python
sequence = pr.PassSequence([
    pass1,
    pass2,
    pass3
])

sequence
```

```python
sequence.solve(in_profile)
```
