```python
import pyroll.basic as pr
```

```python
groove1 = pr.ConstrictedBoxGroove(
    r1 = 1e-9,
    r2 = 1e-3,
    r4 = 2.8e-3,
    depth = 2.1e-3,
    indent = 1.25e-3,
    usable_width = 9e-3,
    ground_width = 8.9e-3
)

groove1
```

```python
profile1 = pr.Profile.from_groove(
    groove = groove1,
    filling = 1,
    gap = 1e-3,
    temperature = 298.15,
    density = 7850,
    specific_heat_capacity = 690,
    freiberg_flow_stress_coefficients = pr.FreibergFlowStressCoefficients(
        a = 900e6,
        m1 = 0,
        m2 = 0.25,
        m3 = 0,
        m4 = 0,
        m5 = 0,
        m6 = 0,
        m7 = 0,
        m8 = 0,
        m9 = 0,
        baseStrain = 1,
        baseStrainRate = 0
    )
)

profile1
```

```python
groove2 = pr.ConstrictedBoxGroove(
    r1 = 1e-9,
    r2 = 1e-3,
    r4 = 2.8e-3,
    depth = 2e-3,
    indent = 1.55e-3,
    usable_width = 10e-3,
    ground_width = 9.9e-3
)

groove2
```

```python
roll = pr.Roll(
    groove = groove2,
    nominal_radius = 15e-3,
    rotational_frequency = 0.318
)

roll
```

```python
pass1 = pr.RollPass(
    roll = roll,
    gap = 1e-3,
    rotation = 0
)

pass1
```

```python
pass1.solve(profile1)
```
