```python
import pyroll.basic as pr
```

```python
in_profile = pr.Profile.round(
    diameter = 60e-3,
    temperature = 298.15,
    density = 7850,
    specific_heat_capacity = 690,
    
    freiberg_flow_stress_coefficients = pr.FreibergFlowStressCoefficients(
        a = 200e6,
        m1 = 0,
        m2 = 0,
        m3 = 0,
        m4 = 0,
        m5 = 0,
        m6 = 0,
        m7 = 0,
        m8 = 0,
        m9 = 0,
        baseStrain = 1,
        baseStrainRate = 0
    )
)

in_profile
```

```python
groove = pr.CircularOvalGroove(
    r1 = 1e-3,
    r2 = 60.5e-3,
    usable_width = 80.88e-3
)

groove
```

```python
roll = pr.Roll(
    groove = groove,
    nominal_radius = 155e-3,
    rotational_frequency = 0.57
)

roll
```

```python
pass1 = pr.RollPass(
    roll = roll,
    gap = 6.5e-3,
)

pass1
```

```python
pass1.solve(in_profile)
```
